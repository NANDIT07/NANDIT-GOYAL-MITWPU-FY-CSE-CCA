```python
import numpy as np

# 1. Create a 3x3 NumPy array of all True
arr1 = np.ones((3, 3), dtype=bool)
print(arr1)


# 2. Create an array of 10 evenly spaced values between 5 and 50
arr2 = np.linspace(5, 50, 10)
print(arr2)


# 3. Convert a Python list into a NumPy array
my_list = [1, 2, 3, 4, 5]
arr3 = np.array(my_list)
print(arr3)


# 4. Reverse a 1D NumPy array
arr4 = np.array([1, 2, 3, 4, 5])
reversed_arr = arr4[::-1]
print(reversed_arr)


# 5. Create a 3x3 identity matrix
arr5 = np.eye(3)
print(arr5)


# 6. Create a 4x4 array and extract its first row and last column
arr6 = np.arange(1, 17).reshape(4, 4)

first_row = arr6[0, :]
last_column = arr6[:, -1]

print("Array:\n", arr6)
print("First row:", first_row)
print("Last column:", last_column)


# 7. Create a 4x4 array and extract first two rows and first two columns
arr7 = np.arange(1, 17).reshape(4, 4)

result = arr7[:2, :2]
print(result)


# 8. Perform arithmetic operations on two NumPy arrays element-wise
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)


# 9. Compute the dot product of two NumPy arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

dot_product = np.dot(a, b)
print("Dot product:", dot_product)


# 10. Compute mean, median, and standard deviation
arr10 = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(arr10))
print("Median:", np.median(arr10))
print("Standard deviation:", np.std(arr10))
```
